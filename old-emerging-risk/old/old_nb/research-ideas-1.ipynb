## 🔧 What is the idea?

Instead of **manually labeling** every 10-K document with its emerging risk types (climate, cyber, geopolitical…), you can **write a set of simple rules or heuristics** — called **labeling functions (LFs)** — that assign **noisy but useful labels** to your data.

Then, you use a tool like **[Snorkel](https://snorkel.org/)** to **combine** these noisy signals into **probabilistic labels** that can be used to **train BERT** as if you had clean ground truth.

---

## 🧱 Basic structure

You define **labeling functions**, which are Python functions like:

```python
def lf_contains_climate_keywords(text):
    keywords = ["greenhouse gas", "emissions", "carbon", "TCFD", "net zero"]
    return 1 if any(k in text.lower() for k in keywords) else 0
```

Another LF might look for geopolitical terms, or match against known cyberattacks:

```python
def lf_mentions_sanctions(text):
    return 1 if "sanctions" in text.lower() or "foreign policy" in text.lower() else 0

def lf_mentions_data_breach(text):
    return 1 if "data breach" in text.lower() or "unauthorized access" in text.lower() else 0
```

Each function returns a **label for one class**, or `-1` if it abstains.

---

## 🧠 What Snorkel does

Snorkel:

1. **Runs all LFs** across your unlabeled data (e.g., 10-K paragraphs or documents)
2. **Builds a probabilistic model** of which LFs are more accurate, correlated, or noisy
3. **Generates soft labels** (e.g., 90% chance this 10-K relates to "cyber risk")
4. These soft labels can be used to **fine-tune BERT** (or any other classifier)

---

## 🧪 Example in context

Imagine your corpus has 10,000 10-Ks. You write \~10–20 labeling functions that detect:

* climate risk keywords or phrases
* cyber risk terminology (e.g., “ransomware”, “SOC”, “penetration test”)
* geopolitical terms (e.g., “tariffs”, “military escalation”, “sanctions”)
* references to disasters (matched to dates from EM-DAT or NOAA)

Then:

* Snorkel generates **noisy but scalable labels** from these LFs
* You now have **10K weakly labeled documents** for **multi-label fine-tuning of BERT**

---

## ✅ Advantages

* **Avoids huge manual annotation** effort
* Leverages **your expert domain knowledge** via interpretable rules
* Enables you to bootstrap a good BERT classifier from scratch
* Supports **multi-label setup** naturally
* Can be updated or refined as you discover new signals

---

## 🧠 Related insight

Snorkel is most useful when:

* You have lots of text (e.g., many 10-Ks)
* Manual labeling is slow
* You can define keyword dictionaries or weak rules
* You plan to use a **strong model** like BERT that benefits from more data, even if noisy

## 📌 What is Zero-Shot Classification?

Zero-shot classification treats classification as a **natural language inference (NLI)** problem:

> “Given a paragraph (the premise), and a label like *This paragraph is about cybersecurity risk* (the hypothesis), how likely is the hypothesis entailed by the paragraph?”

NLI models output:

* **Entailment** (yes, the label fits)
* **Contradiction** (no, it contradicts)
* **Neutral** (uncertain)

This makes the method **flexible and adaptive**, especially to **new risk types not present in training data**.

---

## 🛠️ How It Works

### 1. **Define the risk types as labels**

For example:

```python
risk_labels = [
    "climate risk",
    "cybersecurity risk",
    "geopolitical risk",
    "AI-related risk",
    "supply chain risk",
    "pandemic risk",
    "water scarcity risk"
]
```

### 2. **Load a pretrained zero-shot model**

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
```

### 3. **Classify a paragraph from a 10-K**

```python
text = """Our operations rely heavily on global data infrastructure and may be disrupted by unauthorized access, phishing attacks, or malware."""
result = classifier(text, candidate_labels=risk_labels, multi_label=True)
```

Output:

```python
{
  'labels': ['cybersecurity risk', 'AI-related risk', 'geopolitical risk', ...],
  'scores': [0.92, 0.35, 0.10, ...]
}
```

> The model says: “This paragraph is *very likely* about cybersecurity risk.”

### 4. **Apply across a corpus**

* Chunk each 10-K into paragraphs (or sections)

* Aggregate predictions into document-level risk vectors (e.g., max or average probability per label)
